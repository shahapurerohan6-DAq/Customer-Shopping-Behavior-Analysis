```python
# fill all the null values found in the Review Rating column 
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
#converting all columns into lower case and replacing space with "_" this process we can call it as snake casing 
df.columns=df.columns.str.lower()
df
df.columns=df.columns.str.replace(" ","_")
df=df.rename(columns={"purchase_amount_(usd)":"purchese_amount"})
df
```

```python
# creatin a new column called as age group 
labels=["Young Adult","Adult","Middel","Scenior"]
df["age_group"]=pd.qcut(df["age"],q=4,labels=labels)
df
df[["age_group","age"]].head()
```

```python
"""creating new column called as purchese_frequency_days"""
frequency_mapping = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Monthly" : 30,
    "Quarterly" : 90,
    "Bi-Weekly" : 14,
    "Annually" : 365,
    "Every 3 Months" : 90
}
# df["frequency_of_purchases"]=df["frequency_of_purchases"].str.strip()

df["perchase_frequency_days"]=df["frequency_of_purchases"].map(frequency_mapping)

df
```
